```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
    'Days' : ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    'Study_Hours' : [1, 2, 3, 5, 4],
    'Marks' : [75, 80, 85, 90, 95],
    'Attendance' : [90, 85, 80, 75, 70],
    'Sleep_Hours' : [8, 7, 6, 5, 4],
    'Assignments_Completed' : [5, 4, 3, 2, 1]
}

df = pd.DataFrame(data)
df
```

```python
# 1. Line Chart (Trend)

plt.plot(df['Days'], df['Study_Hours'], marker='o')
plt.title('Study Hours Over Days')
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.show()
```

```python
# Advanced Line Chart
plt.figure(figsize = (7,4))

plt.plot(df['Days'], df['Marks'], marker='o', color = 'red' , label='Marks')

plt.plot(df['Days'], df['Attendance'], marker='o', color = 'green' , label='Attendance')

plt.title('Marks and Attendance Over Days')
plt.xlabel('Days')
plt.ylabel('Marks/Attendance')
plt.legend()
plt
```

```python
# 2. Bar Chart (Comparison)

plt.bar(df['Days'], df['Marks'], label='Marks')

plt.title('Marks by Day')
plt.xlabel('Days')
plt.ylabel('Marks')
plt.legend()
plt
```

```python
#Advanced Bar Graph
bars = plt.bar(df['Days'], df['Marks'], color='red')
for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, y, str(y),
             ha='center', va='bottom')

plt.title("Marks Comparison")
plt.xlabel("Days")
plt.ylabel("Marks")
plt.grid(axis='y')   # adds horizontal grid for readability
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Days' : ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    'Study_Hours' : [1, 2, 3, 5, 4],
    'Marks' : [75, 80, 85, 90, 95],
    'Attendance' : [90, 85, 80, 75, 70],
    'Sleep_Hours' : [8, 7, 6, 5, 4],
    'Assignments_Completed' : [5, 4, 3, 2, 1]
}
df = pd.DataFrame(data)

x = np.arange(len(df['Days']))   # positions for bars
width = 0.35                     # width of each bar

plt.bar(x - width/2, df['Study_Hours'], width, label='Study Hours')
plt.bar(x + width/2, df['Marks'], width, label='Marks')

plt.xticks(x, df['Days'])

plt.title("Study Hours vs Marks")
plt.xlabel("Days")
plt.ylabel("Values")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Example data (more values for better histogram)
marks = [50, 60, 55, 70, 65, 75, 80, 85, 90, 45, 68, 72, 77, 82, 88]

# Calculate mean
mean_value = np.mean(marks)

plt.figure(figsize=(7,4))

plt.hist(marks, bins=6, color='skyblue', edgecolor='black')

# Mean line
plt.axvline(mean_value, color='red', linestyle='--', linewidth=2, label='Mean')

plt.title("Advanced Histogram of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
#3.Histogram (Distribution)
#Shows how values are spread

data = {
    'Marks': [50, 60, 55, 70, 65, 75, 80, 85, 90, 45, 68, 72, 77, 82, 88]
}
df = pd.DataFrame(data)
plt.hist(df['Marks'], bins=5, edgecolor='black',alpha=0.30)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
#4. Scatter Plot (Relationship)
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Days' : ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    'Study_Hours' : [1, 2, 3, 5, 4],
    'Marks' : [75, 80, 85, 90, 95],
    'Attendance' : [90, 85, 80, 75, 70],
    'Sleep_Hours' : [8, 7, 6, 5, 4],
    'Assignments_Completed' : [5, 4, 3, 2, 1]
}
df = pd.DataFrame(data)

plt.scatter(df['Study_Hours'], df['Marks'])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# Conditional scatter plot
# Example category
df['Result'] = ['Fail','Pass','Fail','Pass','Pass']
colors = ['red' if r=='Fail' else 'green' for r in df['Result']]
plt.scatter(df['Study_Hours'], df['Marks'], c=colors)

plt.title("Study Hours vs Marks (Result-wise)")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```
